# Sample Jupyter Notebook with SQL Queries

This notebook demonstrates how to execute SQL queries inside a Jupyter Notebook using SQLite.

```python
# Install required packages (if not installed)
!pip install ipython-sql sqlalchemy sqlite3

# Load SQL extension
%load_ext sql

# Connect to an SQLite database (or create one if it doesn't exist)
%sql sqlite:///sample.db
```

```python
# Execute SQL and store results in a variable
result = %sql SELECT name, age FROM users WHERE age > 30;

# Convert the result to a Pandas DataFrame
df = result.DataFrame()

# Print DataFrame
print(df)

# Iterate over the results
for row in result:
    print(f"User: {row.name}, Age: {row.age}")
```

```python
%%sql
-- Retrieve users older than 30
SELECT * FROM users WHERE age > 30;
```

```python
%%sql
-- Count users in each city
SELECT city, COUNT(*) as user_count FROM users GROUP BY city;
```

```python
%%sql
-- Update age of a user
UPDATE users SET age = 28 WHERE name = 'Bob';

-- Verify update
SELECT * FROM users WHERE name = 'Bob';
```

```python
%%sql
-- Delete a user
DELETE FROM users WHERE name = 'Charlie';

-- Verify deletion
SELECT * FROM users;
```
